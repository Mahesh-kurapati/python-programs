```python
#given year is leap year is not
a=int(input("enter any year: "))
if a%4==0 and a%100!=0 or a%400==0:
   print(f"{a} is a leap year")
else:
   print(f"{a} is not a leap year")
```

```python
#Example for match code
ch=int(input("enter a number:"))
match ch:
  case 0:
    print("MONDAY")
  case 1:
    print("TUESDAY")
  case 2:
    print("WEDNESDAY")
  case 3:
    print("THURSDAY")
  case 4:
    print("FRIDAY")
  case 5:
    print("SATURDAY")
  case 6:
    print("SUNDAY")
  case 7:
    print("INVALID")
```

```python
#program that swaps 2 variables without using temporary variable
a=int(input("enter a number1: "))
b=int (input("enter a number2: "))
print("--Before swapping--")
print(f"a= ",a)
print(f"b= ",b)
a,b = b,a
print("--After swapping--")
print(f"a= ",a)
print(f"b= ",b)
```

```python
#program that finds sum of digits of a given number
a=int(input("enter a number: "))
sum=0
while a!=0:
  b=a%10
  sum=sum+b
  a=a//10
print(sum)
```

```python
#program for reversing a number
a=int(input("enter a number: "))
sum=0
while a!=0:
  b=a%10
  sum=sum*10+b
  a=a//10
print(sum)
```

```python
#program that checks given number is prime or not
a=int(input("enter a number:"))
for i in range(2, a // 2 + 1):
    if a % i == 0:
        print(f"{a} is not a prime number")
        break
else:
  print(f"{a} is a prime number")
```

```python
#program that prints prime numbers between two numbers
a=int(input("enter a number: "))
b=int(input("enter a number: "))
for i in range(a,b+1):
  for j in range(2,i):
    if i%j==0:
      break
  else:
    print(i,end=' ')
```
