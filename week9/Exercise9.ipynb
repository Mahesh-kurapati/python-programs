```python
#write a program to sort words in a file and put them in another file
with open("new.txt","r") as f:
  data = f.read()
print(sorted(data))
```

```python
#file should have only lower-case words and write in order
with open("new.txt","r") as f:
  data = f.read().split()
  data_lower=[i.lower() for i in data]
  data_sort = sorted(data_lower)
with open("new1.txt","w") as f1:
  for i in data_sort:
    f1.write(i)
    f1.write(i+"\n")
with open("new1.txt","r") as f2:
  print(f2.read())
```

```python
import pickle
with open("pickle.bin","wb") as f:
  data = [1,2,3,"hello"]
  pickle.dump(data,f)
with open("pickle.bin","rb") as f1:
  data_read = pickle.load(f1)
print(data_read)
```
