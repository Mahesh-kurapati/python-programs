```python
#dictionaries
student = {"name" : "mahesh","branch" : "AIDS","age" : 19}
print(student["name"])    #accessing
student["email"] = "mahesh@1345"    #adding
student["age"] = 20    #modifying
print(student)
```

```python
#built-in functions
d = {"name" : "mahesh","branch" : "AIDS","age" : 19}
print(len(d))
print(sorted(d))
print(max(d))
```

```python
#dictionary methods
l = {"name" : "mahesh" , "age" : 19}
print(l.keys())
print(l.values())
print(l.items())
l.update({"age" : 20})
print(l)
print(l.pop("name"))
print(l.clear())
```

```python
#program to check if a given key exits in dictionary or not
l = {"name" : "mahesh" , "age" : 19}
key = input()
if key in l:
  print("YES")
else:
  print("NO")
```

```python
#add new value pair to the existing dictionary
l = {"name" : "mahesh" , "age" : 19}
key = input()
value = input()
l[key] = value
print(l)
```

```python
#program to sum the all values in dictionary
a = {1:10 , 2:20 , 3:30}
print(sum(a.values()))
```
