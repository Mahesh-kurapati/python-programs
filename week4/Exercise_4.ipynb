```python
#count no of vowels in a string
a = str(input())
count = 0
for i in a:
  if i=='a' or i=='e' or i=='o' or i=='i' or i=='u' or i=='A' or i=='O' or i=='E' or i=='I' or i=='U':
    count+=1
print(count)
```

```python
#program that prints the reverse of a number using functions
def reversed(a):
  rev = 0
  while(a != 0):
    key = a%10
    rev = rev*10+key
    a = a//10
  return rev
print(reversed(1234))
```
