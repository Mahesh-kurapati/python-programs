```python
#tuple Operations
t1 = (1,2,3,"mahesh","purna")
t2 = (11,12,13)
print(t1 + t2)  #concatenation
print(t2 * 3)   #repetition
print(len(t1))  #length
if t1 not in t2:#membership
  print("YES")
else:
  print("NO")
for i in t1:   #iteration
  print(i)
```

```python
#tuple methods
print(t1.index(2))
print(t2.count(11))
```

```python
print(tuple(zip(t1,t2)))
```

```python
#write a program to create tuples (name, age, address, college) for at least two members and concatenate the tuples and print the concatenated tuples
S1 = tuple(input().split())
S2 = tuple(input().split())
print(S1 + S2)
```

```python
#write a program to print no of vowels in a string(no control flow statements)
s1 = input("Enter a string: ")
s2 = s1.lower()
vowel_count=s2.count("a")+s2.count("e")+s2.count("i")+s2.count("o")+s2.count("u")
print(f"The no of vowels are {vowel_count}")
```
