```python
#list Operations
L1 = [1,2,3,4,5]
L2 = ["HELLO","MAHESH"]
print(L1 + L2) #concatenation
print(L1*3)    #Repetition
print(L2*2)
print(L1[4])   #indexing
print(L2[1])
print(L1[:3])  #Slicing
print(L1[1:2])
if(L1 not in L2):
  print("yes")
else:
  print("no")  #Membership Operator
for i in range(len(L2)):
  print(i)
for i in L1:
  print(i)
```

```python
#Built in Functions For List
L1 = [1,2,3,4,5,6]
L2 = [77,88,100]
L3 = [ 53,25,4,100]
print(len(L2))   #len
print(sum(L3))   #sum
print(min(L1))   #min
print(max(L1))   #max
print(list(reversed(L3))) #reversed
print(sorted(L3))   #sorted
```

```python
#List Methods
L1 = [1,4,2,3,45,5]
L2 = [66,7,33,99,0,100]
#append
L1.append(6)
print(L1)
#extend
L1.extend(L2)
print(L1)
#insert
L1.insert(6,10)
print(L1)
#pop
L1.pop()
print(L1)
#remove
L1.remove(10)
print(L1)
#sort
L1.sort()
print(L1)
#reverse
L1.reverse()
print(L1)
#count
print(L1.count(4))
#index
print(L1.index(1))
```

```python
#program that prints sum of list elements without using sum()
L1 = [2,4,6,8,10,12,14]
sum=0
for i in range(len(L1)):
  sum = sum + L1[i]
print(sum)
```

```python
#print list of elements at odd index
L1 = [2,4,6,8,10,12,14]
for i in range(1,len(L1)):
  if(i%2 != 0):
    print(L1[i])
```

```python
#print the even elements
L1 = [2,4,6,8,10,12,14]
for i in range(len(L1)):
  if(i%2 == 0):
    print(L1[i])
```
