```python
#example on default arguments
def print_msg(name,branch = "AI&DS"):
  print("the student name is",name,"and the branch is",branch)
print_msg("sai")
```

```python
#example on keyword arguments
def subtraction(a,b,c):
  print("The values are ",a,b,c)
  return a-b-c
res = subtraction(a=7,c=9,b=6)
print(res)
```

```python
#program that returns multiple values
def multiple_function(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_function(1,2,3,4,5)
print("the sum is",res_sum,"the count is",res_count,"the average is",res_avg)
```
