```python
#arithematic operators
x=int(input("enter the value of x: "))
y=int(input("enter the value of y: "))
print("sum of two elements",x+y)
print("subtraction of two elements",x-y)
print("multiplication of two elements",x*y)
print("division of two elements",x/y)
print("modulus of two elements",x%y)
print("floor division",x//y)
print("exponent",x**y)
```

```python
#comparsion operators
x=int(input("enter the value of x: "))
y=int(input("enter the value of y: "))
print(x==y)
print(x!=y)
print(x>y)
print(x<y)
print(x>=y)
print(x<=y)
```

```python
#logical operators
x=int(input("enter the value of x: "))
y=int(input("enter the value of y: "))
print(x>y and x<y)
print(x>y or x<y)
print(not(x>y and x<y))
```

```python
#identity operators
x=int(input("enter the value of x: "))
y=int(input("enter the value of y: "))
print(x is y)
print(x is not y)
```

```python
#Membership opertors
x=str(input("enter the value of x: "))
y=str(input("enter the value of y: "))
print(x in y)
print(x not in y)
```

```python
#bitwise operators
x=int(input("enter the value of x: "))
y=int(input("enter the value of y: "))
print(x & y)
print(x | y)
print(x ^ y)
print(x ~ y)
```
