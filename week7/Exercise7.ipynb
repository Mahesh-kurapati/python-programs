```python
#set operations
s1 = {1,2,3,4,5,'mahesh'}
s2 = {3,4,5,7,8,'purna'}
print(s1|s2)
print(s1&s2)
print(s1-s2)
print(s1^s2)
```

```python
#set methods
s1 = {1,2,3,4,5,'mahesh'}
s1.add(10)
print(s1)
s1.update((7,8,9))
print(s1)
s1.remove(10)
print(s1)
s1.discard(1)
print(s1)
s1.pop()
print(s1)
s3=s1.clear()
print(s3)
print(s1)
```

```python
#program to find no of duplicates in a list
l = list(map(int, input().split()))
s = set()
count = 0
for i in l:
    if i in s:
        print(i, end=" ")
        count += 1
    else:
        s.add(i)
print(count)
```

```python
#program to find how many numbers are having duplicates
L = list(map(int, input().split()))
L2 = []
for i in L:
  if L.count(i)>1:
    L2.append(i)
print(L2)
```

```python
#program that prints missing numbers between 1 to n
L = list(map(int,input().split()))
n = int(input())
s = set(L)
for i in range (1,n+1):
  if i not in s:
    print(i,end=" ")
```
